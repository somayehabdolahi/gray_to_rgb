```python
import torch
import matplotlib.pyplot as plt
import torch, numpy,matplotlib
torch.manual_seed(10)
x = torch.linspace(0,1,100).unsqueeze(1)
print(x)
y= 2*x + 1 + torch.randn(x.size())
```

```python

plt.scatter(x, y, label=' Data asli')
plt.title('Simulated Linear Data')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
W = torch.randn(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
W = torch.randn(1,1, requires_grad=True)
b= torch.zeros(1,requires_grad=True)
epochs = 100
losses = []
lr = 0.01
for epoch in range(100):
  y_predicted = x @ W + b
  loss = torch.mean((y_predicted - y)**2)
  losses.append(loss.item())
  loss.backward()
  with torch.no_grad():
    W -= lr*W.grad
    b -= lr*b.grad
  W.grad.zero_()
  b.grad.zero_()
  print("Learned W:", W.item())
  print("Learned b:", b.item())
```

```python
plt.plot(losses)
plt.title('Loss Function')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.show()
with torch.no_grad():
    y_pred_final = W * x + b
plt.scatter(x.detach(), y.detach(), label='Real Data', color='blue')
plt.plot(x.detach(), y_pred_final.detach(), label='Fitted Line', color='red')
plt.title('Learned Linear Model')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
